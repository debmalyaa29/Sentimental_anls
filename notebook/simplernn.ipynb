```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense,Flatten,Input
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
## the dataaset is already vecotrised
X_train.shape
```

```python
##padding
from keras.utils import pad_sequences
X_train = pad_sequences(X_train, maxlen=200, padding='post')
X_test = pad_sequences(X_test, maxlen=200, padding='post')
```

```python
X_train.shape
```

```python
model = Sequential([
    Input(shape=(200,)),
    Embedding(10000, 128),
    SimpleRNN(64, activation='relu',return_sequences=False),
    Dense(1, activation='sigmoid')
])
model.summary()
```

```python
print(X_train.shape)
print(X_test.shape)
print(X_train.dtype)
print(X_test.dtype)
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping
earlystopping=EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
earlystopping
```

```python
## Train the model with early sstopping
history=model.fit(
    X_train,y_train,epochs=10,batch_size=32,
    validation_split=0.2,
    callbacks=[earlystopping]
)
```

```python
## Save model file
model.save('simple_rnn_imdb.h5')
```

```python
loss, accuracy = model.evaluate(X_test, y_test)
```

```python
print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```
