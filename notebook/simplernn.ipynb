```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense,Flatten,Input
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
## the dataaset is already vecotrised
X_train.shape
```

```python
##padding
from keras.utils import pad_sequences
X_train = pad_sequences(X_train, maxlen=200, padding='post')
X_test = pad_sequences(X_test, maxlen=200, padding='post')
```

```python
X_train.shape
```

```python
model = Sequential([
    Input(shape=(200,)),
    Embedding(10000, 128),
    SimpleRNN(64, return_sequences=False),
    Dense(1, activation='sigmoid')
])
model.summary()
```

```python
print(X_train.shape)
print(X_test.shape)
print(X_train.dtype)
print(X_test.dtype)
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
model.fit(X_train,y_train,epochs=15,validation_data=(X_test,y_test))
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2
)
```

```python
loss, accuracy = model.evaluate(X_test, y_test)
```

```python
print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```
