```python
import numpy as np
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense,Flatten,Input

sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good',]
```

```python
##first dtep is to tokenize the sentence
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(oov_token="random")
##oov token helps for any new word convert it to the word random
```

```python
tokenizer.fit_on_texts(sent)
```

```python
tokenizer.word_index
```

```python
##frequency count
tokenizer.word_counts
```

```python
#for findin the rows in sentence
tokenizer.document_count
```

```python
##convert the sentence to vector
sequence=tokenizer.texts_to_sequences(sent)
sequence
```

```python
##applying padding
from keras.utils import pad_sequences
```

```python
sequence=pad_sequences(sequence,padding='post')
```

```python
sequence
```

```python
len(tokenizer.word_index)
```

```python
model = Sequential([
     Input(shape=(5,)),
    Embedding(20, output_dim=6),
    SimpleRNN(64, return_sequences=False),
    Dense(1, activation='sigmoid')
])

model.summary()

```

```python
model.compile('adam','accuracy')
```

```python
pred=model.predict(sequence)
print(pred)
```

```python
classes = (pred >= 0.5).astype(int)

print(classes)
```
